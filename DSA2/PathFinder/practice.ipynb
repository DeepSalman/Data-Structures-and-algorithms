```python
COLORS = ["Red","Green","Blue","Yellow"]
EDGES = [
    ("A", "B"),
    ("A", "C"),
    ("A", "D"),
    ("B", "C"),
    ("B", "E"),
    ("B", "F"),
    ("C", "D"),
    ("C", "E"),
    ("C", "F"),
    ("D", "E"),
    ("D", "G"),
    ("E", "F"),
    ("E", "G"),
    ("E", "H"),
    ("F", "G"),
    ("F", "H"),
    ("G", "H"),
]
VARIABLES = sorted(set(x for edges in EDGES for x in edges))

DOMAINS = {
    variable:COLORS.copy()
    for variable in VARIABLES
}


def build_neighbour(edges):
    """it creates a dictionary where every variable has an empty set as its value."""
    neighbors = {v:set() for v in VARIABLES}
    neighbors = {v:set() for v in VARIABLES}

    for x,y in edges:
        neighbors[x].add(y)
        neighbors[y].add(x)

    return neighbors

def build_neighbors(edges):
    neighbors = {v:set() for v in VARIABLES}

    for x,y in edges:
        neighbors[x].add(y)
        neighbors[y].add(x)

    return neighbors




NEIGHBORS = build_neighbour(EDGES)


def is_consistent(variable,value,assignment):
    for neighbors in NEIGHBORS[variable]:
        if neighbors in assignment:
            if assignment[neighbors]==value:
                return False
    return True

def is_consistent(variable,value,assignment):
    for neighbors in NEIGHBORS[variable]:
        if neighbors in assignment:
            if assignment[neighbors]==value:
                return False
    return True

def is_consistent(variable,value,assignment):
    for neighbors in NEIGHBORS[variable]:
        if neighbors in assignment:
            if assignment[neighbors]==value:
                return False 


    return True


def basic_backtracking(assignment,domains):
    if len(assignment)==len(VARIABLES):
        return assignment.copy()

    variables= next(v for v in variables
                    if v not in assignment)

    for value in domains[variables]:
        if is_consistent(variables,value,assignment):
            assignment[variables]=value
            result = basic_backtracking(assignment,domains)

            if result is not None:
                return result

            del assignment[variables]

    return None


def forward_check(variable,value,assignment,domains):
    new_domains=deepcopy(domains)
    new_domains[variable]=[value]

    for neighbors in NEIGHBORS[variable]:
        if neighbors in assignment:
            continue

        if value in new_domains[neighbors]:
            new_domains[neighbors].remove(value)

        if len(new_domains[neighbors])==0:
            return None

    return new_domains

def forward_check(variable,value,assignment,domain):
    new_domain = deepcopy(domain)
    new_domain[variable]=[value]

    for neighbor in NEIGHBORS[variable]:
        if neighbor in assignment:
            continue
        if value in new_domain[neighbor]:
            new_domain[neighbor].remove(value)

        if len(new_domain[neighbor])==0:
            return None
    return new_domain


def revise(domains,x,y):
    revised = False

    for x_value in domains[x][:]:
        supported = any(x_value != y_value
                        for y_value in domains[y])

        if not supported:
            domains[x].remove(x_value)
            revised = True


    return revised 

def revised(domains,x,y):
    revised = False

    for x_values in domains[x][:]:
        supported = any (x_values!=y_values
            for y_values in domains[y])
        if not supported:
            domains[x].remove(x_values)

    return revised 

def revised(domains,x,y):
    revised = False
    for x_values in domains[x][:]:
        supported = any(x_values!=y_values
                        for y_values in domains[y])

        if not supported:
            domains[x].remove(x_values)
            revised = True
    return revised

def ac3(domains):
    queue = deque()
    for x in VARIABLES:
        for y in NEIGHBORS[x]:
            queue.append((x,y))


    while queue:
        x,y = queue.popleft()
        if revised(domains,x,y):
            if len(domains[x]==0):
                return False
            for z in NEIGHBORS[x]:
                if z!=y:
                    queue.append((z,x))

    while queue:

        x,y = queue.popleft()   

        if revise(domains,x,y):
            if len(domains[x])==0:
                return False
            for z in NEIGHBORS[x]:
                if z!=y:
                    queue.append((z,x))


    return True


def select_mrv_variable(assignment,domains):
    unassigned = [v for v in VARIABLES
                  if v not in assignment]

    return min(unassigned,key=lambda v:len(domains[v]))

def improved_backtracking(assignment,domain):
    if len(assignment)==len(VARIABLES):
        return assignment.copy()

    variables = select_mrv_variable(assignment,domain)


    for value in domain[variables]:
        if not is_consistent(variables,value,assignment):
            continue

        new_domain= forward_check(variables,value,assignment,domain)

        if new_domain is None:
            continue
        if not ac3(new_domain):
            continue

        



        assignment[variables]=value

        result = improved_backtracking(assignment,domain)

        if result is not None:
            return result

        del assignment[variables]




    



```
